```python
#works better
import os
import cv2
import numpy as np
import logging
from pdf2image import convert_from_path
from PIL import Image
from IPython.display import display

# =============================================================================
# Logger Manager
# =============================================================================
class LoggerManager:
    """
    A dedicated logger manager to configure a global logger.
    The logger is set up to provide detailed trace information both on the console and in a log file.
    """

    @staticmethod
    def configure_global_logger(debug: bool = False, log_file: str = "logs/application.log") -> logging.Logger:
        """
        Configure and return a global logger instance.
        
        Args:
            debug (bool): If True, set logger level to DEBUG; otherwise INFO.
            log_file (str): Path to the log file where detailed logs will be stored.
            
        Returns:
            logging.Logger: A configured logger instance.
        """
        os.makedirs(os.path.dirname(log_file), exist_ok=True)
        logger = logging.getLogger("BlueBoxProcessing")
        logger.setLevel(logging.DEBUG if debug else logging.INFO)
        # Remove any previously attached handlers.
        for handler in logger.handlers[:]:
            logger.removeHandler(handler)
        formatter = logging.Formatter("%(asctime)s %(levelname)-8s [%(name)s] %(message)s")
        
        # Create file handler for detailed logging.
        fh = logging.FileHandler(log_file)
        fh.setLevel(logging.DEBUG)
        fh.setFormatter(formatter)
        logger.addHandler(fh)
        
        # Create stream handler for real-time console output.
        ch = logging.StreamHandler()
        ch.setLevel(logging.INFO)
        ch.setFormatter(formatter)
        logger.addHandler(ch)
        
        logger.debug("Global logger configured with log file: %s. Debug mode set to %s", log_file, debug)
        return logger

# Initialize the global logger.
global_logger = LoggerManager.configure_global_logger(debug=False)

# =============================================================================
# Utility Functions
# =============================================================================
def dbfs_to_local_path(dbfs_path):
    """
    Convert a Databricks File System (DBFS) URI into a local filesystem path.
    
    For example, a DBFS URI like "dbfs:/mnt/..." becomes "/dbfs/mnt/...".
    
    Args:
        dbfs_path (str): The DBFS URI string.
        
    Returns:
        str: The corresponding local filesystem path.
    """
    global_logger.debug("dbfs_to_local_path: Received path '%s'.", dbfs_path)
    if dbfs_path.startswith("dbfs:/"):
        local_path = "/dbfs/" + dbfs_path[len("dbfs:/"):]
        global_logger.debug("dbfs_to_local_path: Converted to local path '%s'.", local_path)
        return local_path
    global_logger.debug("dbfs_to_local_path: No conversion necessary for path '%s'.", dbfs_path)
    return dbfs_path


def convert_pdf_to_images(pdf_path, dpi=300):
    """
    Convert each page of a PDF document to a list of high-resolution PIL Image objects.
    
    Args:
        pdf_path (str): The filesystem path to the PDF file.
        dpi (int): The dots-per-inch (DPI) resolution for conversion (default is 300).
        
    Returns:
        list: List of PIL.Image.Image instances, one per page.
        
    Logging:
        Logs the PDF path, DPI, and number of pages converted.
    """
    global_logger.info("convert_pdf_to_images: Converting PDF '%s' into images using DPI=%d.", pdf_path, dpi)
    pages = convert_from_path(pdf_path, dpi=dpi)
    global_logger.debug("convert_pdf_to_images: Successfully converted %d pages.", len(pages))
    return pages


def show_image_in_notebook(pil_image, title=""):
    """
    Display a PIL image inline within the notebook.
    
    Args:
        pil_image (PIL.Image.Image): The image to be displayed.
        title (str): An optional title for the image display.
        
    Logging:
        Logs that the image is being displayed with an optional title.
    """
    global_logger.info("show_image_in_notebook: Displaying image. Title: '%s'.", title)
    display(pil_image)


def detect_small_blue_boxes(
    pil_image,
    lower_blue=np.array([100, 100, 50]),
    upper_blue=np.array([130, 255, 255]),
    min_area=1000,
    max_area=200000,
    max_width=None,
    max_height=200
):
    """
    Detect distinct blue boxes within a given PIL image based on HSV thresholding and morphological processing.
    
    The function performs the following operations:
      1. Converts the PIL image to an OpenCV BGR image.
      2. Converts the image to HSV color space and applies a threshold to isolate blue regions.
      3. Applies a small morphological opening to reduce noise.
      4. Optionally applies a morphological closing to fill tiny gaps.
      5. Uses connected components analysis to find distinct regions.
      6. Filters these regions based on area and size constraints.
    
    Args:
        pil_image (PIL.Image.Image): The input image.
        lower_blue (np.array): Lower boundary for blue color in HSV space.
        upper_blue (np.array): Upper boundary for blue color in HSV space.
        min_area (int): Minimum acceptable area for a detected box.
        max_area (int): Maximum acceptable area for a detected box.
        max_width (int or None): Maximum acceptable width (None means no limit).
        max_height (int): Maximum acceptable height for a valid box.
        
    Returns:
        list: A list of tuples (x, y, w, h) representing bounding boxes.
    
    Logging:
        Provides detailed logs about conversion, shape sizes, applied kernels, and box filtering.
    """
    global_logger.info("detect_small_blue_boxes: Starting blue box detection.")
    
    # Convert PIL to a NumPy array (RGB) then to BGR for OpenCV.
    img_rgb = np.array(pil_image.convert("RGB"))
    global_logger.debug("detect_small_blue_boxes: Converted PIL image to RGB array of shape %s.", img_rgb.shape)
    img_bgr = cv2.cvtColor(img_rgb, cv2.COLOR_RGB2BGR)
    global_logger.debug("detect_small_blue_boxes: Converted RGB image to BGR.")

    # Convert image to HSV and threshold for the blue color range.
    hsv = cv2.cvtColor(img_bgr, cv2.COLOR_BGR2HSV)
    mask = cv2.inRange(hsv, lower_blue, upper_blue)
    global_logger.debug("detect_small_blue_boxes: Applied HSV thresholding. Mask shape: %s.", mask.shape)

    # Perform morphological open to remove noise.
    kernel_open = cv2.getStructuringElement(cv2.MORPH_RECT, (2, 2))
    global_logger.debug("detect_small_blue_boxes: Using morphological open kernel of shape %s.", kernel_open.shape)
    mask_opened = cv2.morphologyEx(mask, cv2.MORPH_OPEN, kernel_open, iterations=1)
    global_logger.debug("detect_small_blue_boxes: Completed morphological opening.")

    # Perform a morphological close to fill small gaps.
    kernel_close = cv2.getStructuringElement(cv2.MORPH_RECT, (2, 2))
    global_logger.debug("detect_small_blue_boxes: Using morphological close kernel of shape %s.", kernel_close.shape)
    mask_closed = cv2.morphologyEx(mask_opened, cv2.MORPH_CLOSE, kernel_close, iterations=1)
    global_logger.debug("detect_small_blue_boxes: Completed morphological closing.")

    # Connected components to detect individual regions.
    num_labels, labels = cv2.connectedComponents(mask_closed)
    global_logger.debug("detect_small_blue_boxes: Found %d connected components (ignoring background).", num_labels - 1)

    boxes = []
    # Loop over each component, skipping the background (label 0).
    for label in range(1, num_labels):
        ys, xs = np.where(labels == label)
        x_min, x_max = np.min(xs), np.max(xs)
        y_min, y_max = np.min(ys), np.max(ys)
        w = x_max - x_min + 1
        h = y_max - y_min + 1
        area = w * h

        global_logger.debug("detect_small_blue_boxes: Label %d has bounding box (x=%d, y=%d, w=%d, h=%d, area=%d).", 
                              label, x_min, y_min, w, h, area)
        # Filter boxes based on area and size constraints.
        if area < min_area:
            global_logger.debug("Label %d filtered out due to small area (%d < %d).", label, area, min_area)
            continue
        if area > max_area:
            global_logger.debug("Label %d filtered out due to large area (%d > %d).", label, area, max_area)
            continue
        if h > max_height:
            global_logger.debug("Label %d filtered out due to excessive height (%d > %d).", label, h, max_height)
            continue
        if max_width is not None and w > max_width:
            global_logger.debug("Label %d filtered out due to excessive width (%d > %d).", label, w, max_width)
            continue

        boxes.append((x_min, y_min, w, h))
        global_logger.debug("Label %d accepted with bounding box: (x=%d, y=%d, w=%d, h=%d).", 
                              label, x_min, y_min, w, h)

    # Sort boxes top-to-bottom then left-to-right.
    boxes.sort(key=lambda b: (b[1], b[0]))
    global_logger.info("detect_small_blue_boxes: Detection complete with %d valid blue boxes.", len(boxes))
    return boxes


def filter_header_boxes(boxes, target_height=31, tolerance=5):
    """
    Filter a list of bounding boxes to retain only those that meet a specified header height criterion.
    
    Args:
        boxes (list): List of bounding boxes in (x, y, w, h) format.
        target_height (int): The desired height for a header box.
        tolerance (int): Allowed deviation in height from the target.
        
    Returns:
        list: Filtered list of boxes that qualify as header boxes.
    
    Logging:
        Logs details of each box's height comparison.
    """
    global_logger.info("filter_header_boxes: Filtering boxes to find header boxes with target height %d ± %d.", target_height, tolerance)
    header_boxes = []
    for (x, y, w, h) in boxes:
        if abs(h - target_height) <= tolerance:
            header_boxes.append((x, y, w, h))
            global_logger.debug("filter_header_boxes: Box (x=%d, y=%d, w=%d, h=%d) accepted.", x, y, w, h)
        else:
            global_logger.debug("filter_header_boxes: Box (x=%d, y=%d, w=%d, h=%d) rejected.", x, y, w, h)
    header_boxes.sort(key=lambda b: (b[1], b[0]))
    global_logger.info("filter_header_boxes: %d header boxes found.", len(header_boxes))
    return header_boxes


def crop_sections_by_headers(pil_image, headers, row_tolerance=10):
    """
    Crop a page into several sections based on detected header positions.
    
    This function splits the page into a "pre-header" section (from the page top to the first header row)
    and additional sections delimited by header rows. It groups headers that are vertically close.
    
    Args:
        pil_image (PIL.Image.Image): The input image of the page.
        headers (list): A list of header bounding boxes (x, y, w, h).
        row_tolerance (int): Maximum vertical distance (in pixels) to consider boxes as in the same row.
        
    Returns:
        tuple:
            - sections (list): A list of cropped PIL.Image.Image sections.
            - section_coords (list): A list of coordinates (left, top, right, bottom) for each section.
    
    Logging:
        Logs grouping details, row positions, and final cropped section coordinates.
    """
    global_logger.info("crop_sections_by_headers: Starting to crop page into sections based on header positions.")
    width, height = pil_image.width, pil_image.height
    sections = []
    section_coords = []

    if not headers:
        global_logger.warning("crop_sections_by_headers: No headers detected. Entire page is considered one section.")
        sections.append(pil_image)
        section_coords.append((0, 0, width, height))
        return sections, section_coords

    # Group header boxes into rows based on vertical position.
    rows = []
    current_row = [headers[0]]
    for i in range(1, len(headers)):
        x, y, w, h = headers[i]
        _, prev_y, _, prev_h = headers[i - 1]
        if abs(y - prev_y) <= row_tolerance:
            current_row.append(headers[i])
        else:
            rows.append(current_row)
            current_row = [headers[i]]
    if current_row:
        rows.append(current_row)
    global_logger.debug("crop_sections_by_headers: Headers grouped into %d rows.", len(rows))

    # Determine top coordinate for each row and append page bottom.
    row_tops = []
    for row_headers in rows:
        row_top = min(hdr[1] for hdr in row_headers)
        row_tops.append(row_top)
    row_tops.append(height)
    global_logger.debug("crop_sections_by_headers: Row top positions determined as: %s", row_tops)

    # Create a pre-header section for the area above the first row.
    first_row_top = row_tops[0]
    if first_row_top > 0:
        pre_section = pil_image.crop((0, 0, width, first_row_top))
        sections.append(pre_section)
        section_coords.append((0, 0, width, first_row_top))
        global_logger.debug("crop_sections_by_headers: Created pre-header section covering (0,0,%d,%d).", width, first_row_top)

    # Process each header row to define sections.
    for r_idx, row_headers in enumerate(rows):
        this_row_top = row_tops[r_idx]
        next_row_top = row_tops[r_idx + 1]

        # Sort header boxes in the row from left to right.
        row_headers.sort(key=lambda b: b[0])
        for j in range(len(row_headers)):
            x, _, w, _ = row_headers[j]
            left = x
            right = row_headers[j + 1][0] if j < len(row_headers) - 1 else width

            cropped = pil_image.crop((left, this_row_top, right, next_row_top))
            sections.append(cropped)
            section_coords.append((left, this_row_top, right, next_row_top))
            global_logger.debug("crop_sections_by_headers: Cropped section with coordinates (left=%d, top=%d, right=%d, bottom=%d).", left, this_row_top, right, next_row_top)

    global_logger.info("crop_sections_by_headers: Cropped a total of %d sections.", len(sections))
    return sections, section_coords


def crop_and_display_save_boxes(pil_image, boxes, output_folder_dbfs, page_index, prefix):
    """
    Crop detected boxes from the input image, display them inline, and save them to disk.
    
    This function performs the following:
      1. Creates the target output directory on DBFS.
      2. Converts the DBFS path to a local filesystem path.
      3. For each bounding box, crops the corresponding region from the image.
      4. Displays the cropped box in the notebook.
      5. Saves the cropped box image with a filename that includes the page index and a prefix.
    
    Args:
        pil_image (PIL.Image.Image): The input image of the page.
        boxes (list): List of bounding boxes (x, y, w, h).
        output_folder_dbfs (str): DBFS URI where the cropped images will be saved.
        page_index (int): Index of the current page (starting from 0).
        prefix (str): A prefix for the filename (e.g., "Header").
    
    Logging:
        Logs each crop, its coordinates, display actions, and save operations.
    """
    global_logger.info("crop_and_display_save_boxes: Starting to crop, display, and save boxes with prefix '%s'.", prefix)
    # Create the directory on DBFS.
    dbutils.fs.mkdirs(output_folder_dbfs)
    global_logger.debug("crop_and_display_save_boxes: Ensured output folder '%s' exists on DBFS.", output_folder_dbfs)
    # Convert DBFS path to a local path.
    output_folder_local = dbfs_to_local_path(output_folder_dbfs)
    global_logger.debug("crop_and_display_save_boxes: Converted DBFS folder to local folder '%s'.", output_folder_local)

    for i, (x, y, w, h) in enumerate(boxes):
        cropped = pil_image.crop((x, y, x + w, y + h))
        message = f"{prefix} Box {i+1} at (x={x}, y={y}, w={w}, h={h})"
        global_logger.info("crop_and_display_save_boxes: %s", message)
        print(message)
        show_image_in_notebook(cropped, title=f"{prefix} Box {i+1}")
        filename = f"page_{page_index+1}_{prefix.lower().replace(' ', '_')}_box_{i+1}.png"
        save_path = os.path.join(output_folder_local, filename)
        cropped.save(save_path)
        message = f"Saved {prefix} cropped image to: {save_path}\n"
        global_logger.info("crop_and_display_save_boxes: %s", message)
        print(message)


# =============================================================================
# Main Pipeline
# =============================================================================
def main():
    """
    Main pipeline to process a PDF document and extract visual elements.
    
    The pipeline performs the following:
      1. Converts the specified PDF into high-resolution image pages.
      2. For each page:
          a. Detects blue boxes using color thresholding and morphological operations.
          b. Draws bounding boxes for debugging and saves the debug image.
          c. Filters the detected boxes to extract header boxes (approx. 31 pixels in height).
          d. Crops, displays, and saves each header box.
          e. Segments the page into sections based on header locations and saves each section.
    
    Paths and parameters can be adjusted as necessary. Detailed logging provides insight into each step.
    """
    global_logger.info("main: Starting main pipeline for PDF processing.")
    # Define paths (modify for your environment).
    pdf_path = "/dbfs/mnt/mini-proj-dd/DDR.pdf"  # PDF path (assumed accessible)
    output_folder_boxes_dbfs = "dbfs:/mnt/mini-proj-dd/cropped_blue_boxes"
    output_folder_sections_dbfs = "dbfs:/mnt/mini-proj-dd/cropped_sections"
    
    # Convert PDF pages to images using high resolution (dpi = 300 ensures clarity).
    pages = convert_pdf_to_images(pdf_path, dpi=300)
    global_logger.info("main: PDF converted into %d image pages.", len(pages))
    
    for page_index, pil_image in enumerate(pages):
        message = f"\n--- Processing Page {page_index+1} ---"
        global_logger.info("main: %s", message)
        print(message)

        # Step 1: Detect distinct blue boxes.
        boxes = detect_small_blue_boxes(
            pil_image,
            lower_blue=np.array([100, 100, 50]),
            upper_blue=np.array([130, 255, 255]),
            min_area=1000,
            max_area=200000,
            max_width=None,
            max_height=200
        )
        message = f"main: Detected {len(boxes)} blue boxes on Page {page_index+1}."
        global_logger.info(message)
        print(message)

        # Draw bounding boxes on the page for visual debugging.
        debug_img = np.array(pil_image.convert("RGB"))
        for (x, y, w, h) in boxes:
            cv2.rectangle(debug_img, (x, y), (x + w, y + h), (255, 0, 0), 2)
        debug_pil = Image.fromarray(debug_img)
        message = "main: Displaying bounding boxes on the page:"
        global_logger.info(message)
        print(message)
        show_image_in_notebook(debug_pil, title="Detected Blue Boxes")

        # Save the debug image showing blue boxes.
        dbutils.fs.mkdirs(output_folder_boxes_dbfs)
        output_folder_boxes_local = dbfs_to_local_path(output_folder_boxes_dbfs)
        boxes_image_path = os.path.join(output_folder_boxes_local, f"page_{page_index+1}_blue_boxes.png")
        debug_pil.save(boxes_image_path)
        message = f"main: Saved blue boxes image to: {boxes_image_path}\n"
        global_logger.info(message)
        print(message)

        # Step 2: Filter out header boxes (by height around 31 pixels).
        header_boxes = filter_header_boxes(boxes, target_height=31, tolerance=5)
        message = f"main: Filtered {len(header_boxes)} header boxes on Page {page_index+1}."
        global_logger.info(message)
        print(message)

        # Crop, display, and save each detected header box.
        crop_and_display_save_boxes(pil_image, header_boxes, output_folder_boxes_dbfs, page_index, prefix="Header")

        # Step 3: Crop page sections based on header positions.
        sections, section_coords = crop_sections_by_headers(pil_image, header_boxes, row_tolerance=10)
        message = f"main: Cropped {len(sections)} sections from Page {page_index+1}."
        global_logger.info(message)
        print(message)

        # Display and save each cropped section.
        dbutils.fs.mkdirs(output_folder_sections_dbfs)
        output_folder_sections_local = dbfs_to_local_path(output_folder_sections_dbfs)
        for i, section in enumerate(sections):
            message = f"main: Section {i+1} coordinates: {section_coords[i]}"
            global_logger.info(message)
            print(message)
            show_image_in_notebook(section, title=f"Section {i+1}")
            section_filename = os.path.join(output_folder_sections_local, f"page_{page_index+1}_section_{i+1}.png")
            section.save(section_filename)
            message = f"main: Saved section image to: {section_filename}\n"
            global_logger.info(message)
            print(message)
    
    global_logger.info("main: Pipeline completed successfully. All pages processed.")


# =============================================================================
# Entry Point
# =============================================================================
if __name__ == "__main__":
    main()
```
